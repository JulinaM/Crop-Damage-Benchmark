```python

import tarfile
from rasterio.io import MemoryFile

tar_path = '/users/PGS0218/julina/projects/geography/Crop-Damage-Benchmark/data/input/J_AgDamage/AgDamage_v2/Flood/shards/train/train-000000.tar'
tar = tarfile.open(tar_path, 'r')
names = tar.getnames()
label_names = [n for n in names if n.endswith('.label.tif')]
print('num label members found:', len(label_names))
print('sample:', label_names[0])

raw = tar.extractfile(label_names[0]).read()
with MemoryFile(raw) as mf, mf.open() as src:
    print('count (bands):', src.count)
    print('dtype:', src.dtypes)
    print('nodata:', src.nodata)
    print('shape:', src.height, src.width)
    arr = src.read()
    import numpy as np
    for b in range(src.count):
        vals, counts = np.unique(arr[b], return_counts=True)
        print(f'band {b+1} unique values:', dict(zip(vals.tolist(), counts.tolist())))
```
